## Parte 1: Análisis Exploratorio y Visualización

```python
import pandas as pd
import numpy as np

# 1. Carga del dataset y visualización de las primeras 10 filas
df = pd.read_csv('02. Material de apoyo - Estudiantes_edutech.csv')
display(df.head(10))

# 2. Identificación de nulos y tipos de datos
print("--- Información del Dataset ---")
df.info()

print("\n--- Conteo de Valores Nulos ---")
print(df.isnull().sum())

# 3. Estadísticas descriptivas de variables numéricas
print("\n--- Estadísticas Descriptivas ---")
display(df.describe())

# 4. Correlación usando NumPy
# Se excluye 'id_estudiante' (no predictiva) y
#'tareas_totales'
num_cols = df.select_dtypes(include=[np.number]).columns.drop(['id_estudiante', 'tareas_totales'])
corr_matrix = np.corrcoef(df[num_cols].values, rowvar=False)

# Pasar a df
corr_df = pd.DataFrame(corr_matrix, index=num_cols, columns=num_cols)
print("\n--- Matriz de Correlación (NumPy) ---")
display(corr_df)

# 5. Identificación de Outliers (Método IQR)
def find_outliers_iqr(series):
    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = series[(series < lower_bound) | (series > upper_bound)]
    return outliers

outliers_horas = find_outliers_iqr(df['horas_estudio_semanal'])
outliers_monto = find_outliers_iqr(df['monto_inversion'])

print(f"\n--- Análisis de Outliers ---")
print(f"Outliers en horas_estudio_semanal: {len(outliers_horas)}")
print(f"Outliers en monto_inversion: {len(outliers_monto)}")
```

## Parte 2: Preparación de Datos y Feature Engineering

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, OneHotEncoder, StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from imblearn.over_sampling import SMOTE

print("=== PARTE 2.1: TRANSFORMACIÓN DE VARIABLES ===")

# 1. Creación de nuevas variables
# tasa_completitud = (tareas_completadas / tareas_totales) * 100
df['tasa_completitud'] = (df['tareas_completadas'] / df['tareas_totales']) * 100

# estudiante_activo = 1 si sesiones_totales > 50, sino 0
df['estudiante_activo'] = np.where(df['sesiones_totales'] > 50, 1, 0)
print("Nuevas variables 'tasa_completitud' y 'estudiante_activo' creadas.")

# Se eliminan columnas que no aportan valor predictivo (identificadores)
df = df.drop(['id_estudiante', 'nombre'], axis=1)

# 2. Separación de Features (X) y Targets (y) ANTES de transformar para evitar fugas
X = df.drop(['curso_completado', 'puntaje_satisfaccion'], axis=1)
y_clasificacion = df['curso_completado']
y_regresion = df['puntaje_satisfaccion']

# 3. Identificación de tipos de columnas para el preprocesamiento
# Ordinales (orden lógico)
cols_ordinales = ['nivel_educacion', 'experiencia_previa']
# Nominales (no tienen un orden lógico)
cols_nominales = ['genero', 'pais', 'dispositivo_principal', 'horario_estudio', 'plataforma_origen']
# Numéricas continuas (se normalizarán)
cols_numericas = ['edad', 'horas_estudio_semanal', 'sesiones_totales', 'tareas_completadas',
                  'tareas_totales', 'participacion_foros', 'promedio_evaluaciones',
                  'monto_inversion', 'tasa_completitud']


# 4. Codificación de variables categóricas
# Aplicar LabelEncoder a las ordinales (modifica X directamente)
le = LabelEncoder()
for col in cols_ordinales:
    X[col] = le.fit_transform(X[col])

# Construir un ColumnTransformer para aplicar OneHotEncoder y StandardScaler
# drop='first' en OneHotEncoder evita la multicolinealidad
preprocesador = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), cols_numericas),
        ('nom', OneHotEncoder(drop='first', sparse_output=False), cols_nominales)
    ],
    remainder='passthrough' # Mantiene las ordinales y binarias
)

# Aplicar las transformaciones a X
X_procesado_array = preprocesador.fit_transform(X)

# Recuperar los nombres de las columnas post-OneHotEncoding
nombres_nominales = preprocesador.named_transformers_['nom'].get_feature_names_out(cols_nominales)
nombres_finales = cols_numericas + list(nombres_nominales) + cols_ordinales + ['estudiante_activo']

X_procesado = pd.DataFrame(X_procesado_array, columns=nombres_finales)
print("Transformaciones completadas: LabelEncoder, OneHotEncoder y StandardScaler aplicados.\n")


print("=== PARTE 2.2: DIVISIÓN Y PREPARACIÓN ===")

# 5. División en Entrenamiento (70%) y Prueba (30%)
# Usar random_state=42
X_train, X_test, y_train_clf, y_test_clf, y_train_reg, y_test_reg = train_test_split(
    X_procesado, y_clasificacion, y_regresion, test_size=0.3, random_state=42, stratify=y_clasificacion
)
print(f"División completada. Set de entrenamiento: {X_train.shape[0]} filas. Set de prueba: {X_test.shape[0]} filas.")

# 6. Aplicación de SMOTE (Solo en el set de entrenamiento)
print("\nDistribución de clases ANTES de SMOTE:")
print(y_train_clf.value_counts())

# Inicializar y aplicar SMOTE
smote = SMOTE(random_state=42)
X_train_smote, y_train_clf_smote = smote.fit_resample(X_train, y_train_clf)

print("\nDistribución de clases DESPUÉS de SMOTE:")
print(y_train_clf_smote.value_counts())
print("\n¡Preparación de datos completada con éxito y sin fuga de información (data leakage)!")
```

### 🔄 Transformación
- Se utilizó **LabelEncoder** para las variables **educación** y **experiencia**, ya que poseen una **jerarquía natural**.  
- Para el resto de las variables categóricas se aplicó **OneHotEncoder**, dado que no existe un orden implícito entre sus categorías.  

---

### ✂️ División y SMOTE
- La división **70/30** se realizó de forma **estratificada** para mantener la proporción original de la clase minoritaria.  
- Al observar el desbalanceo (más estudiantes que **NO completaron** el curso que los que sí), se aplicó **SMOTE** estrictamente sobre `X_train`.  
- Esto balancea las clases creando **datos sintéticos** en el conjunto de entrenamiento, pero deja `X_test` intacto para evaluar el modelo en **condiciones reales**.  


## Parte 3: Modelado Predictivo con Machine Learning

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import GridSearchCV
from sklearn.linear_model import LogisticRegression, Ridge
from sklearn.ensemble import RandomForestClassifier, RandomForestRegressor
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, confusion_matrix
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

print("======================================================")
print(" PARTE 3.1: MODELOS DE CLASIFICACIÓN (Curso Completado)")
print("======================================================\n")

# 1. Regresión Logística
print("Entrenando Logistic Regression con GridSearchCV...")
param_grid_lr = {'C': [0.01, 0.1, 1, 10], 'penalty': ['l1', 'l2'], 'solver': ['liblinear']}
grid_lr = GridSearchCV(LogisticRegression(random_state=42), param_grid_lr, cv=5, scoring='f1')
grid_lr.fit(X_train_smote, y_train_clf_smote)
best_lr = grid_lr.best_estimator_
y_pred_lr = best_lr.predict(X_test)
y_prob_lr = best_lr.predict_proba(X_test)[:, 1]

# 2. Random Forest Classifier
print("Entrenando Random Forest Classifier con GridSearchCV...")
param_grid_rfc = {'n_estimators': [50, 100], 'max_depth': [5, 10, None], 'min_samples_split': [2, 5]}
grid_rfc = GridSearchCV(RandomForestClassifier(random_state=42), param_grid_rfc, cv=5, scoring='f1')
grid_rfc.fit(X_train_smote, y_train_clf_smote)
best_rfc = grid_rfc.best_estimator_
y_pred_rfc = best_rfc.predict(X_test)
y_prob_rfc = best_rfc.predict_proba(X_test)[:, 1]

# 3. Evaluación y Comparación
def evaluar_clasificacion(nombre, y_true, y_pred, y_prob, modelo):
    print(f"--- Resultados {nombre} ---")
    print(f"Mejores hiperparámetros: {modelo.get_params()}")
    print(f"Accuracy:  {accuracy_score(y_true, y_pred):.4f}")
    print(f"Precision: {precision_score(y_true, y_pred):.4f}")
    print(f"Recall:    {recall_score(y_true, y_pred):.4f}")
    print(f"F1-Score:  {f1_score(y_true, y_pred):.4f}")
    print(f"ROC-AUC:   {roc_auc_score(y_true, y_prob):.4f}")
    print("Matriz de Confusión:")
    print(confusion_matrix(y_true, y_pred))
    print("\n")

evaluar_clasificacion("Logistic Regression", y_test_clf, y_pred_lr, y_prob_lr, best_lr)
evaluar_clasificacion("Random Forest Classifier", y_test_clf, y_pred_rfc, y_prob_rfc, best_rfc)


print("======================================================")
print(" PARTE 3.2: MODELOS DE REGRESIÓN (Puntaje Satisfacción)")
print("======================================================\n")

# 1. Regresión Ridge (Usar X_train normal, sin SMOTE)
print("Entrenando Ridge Regression con GridSearchCV...")
param_grid_ridge = {'alpha': [0.1, 1.0, 10.0, 100.0]}
grid_ridge = GridSearchCV(Ridge(random_state=42), param_grid_ridge, cv=5, scoring='neg_mean_squared_error')
grid_ridge.fit(X_train, y_train_reg)
best_ridge = grid_ridge.best_estimator_
y_pred_ridge = best_ridge.predict(X_test)

# 2. Random Forest Regressor
print("Entrenando Random Forest Regressor con GridSearchCV...")
param_grid_rfr = {'n_estimators': [50, 100], 'max_depth': [5, 10, None]}
grid_rfr = GridSearchCV(RandomForestRegressor(random_state=42), param_grid_rfr, cv=5, scoring='neg_mean_squared_error')
grid_rfr.fit(X_train, y_train_reg)
best_rfr = grid_rfr.best_estimator_
y_pred_rfr = best_rfr.predict(X_test)

# 3. Evaluación y Comparación
def evaluar_regresion(nombre, y_true, y_pred):
    print(f"--- Resultados {nombre} ---")
    print(f"R²:   {r2_score(y_true, y_pred):.4f}")
    print(f"MAE:  {mean_absolute_error(y_true, y_pred):.4f}")
    print(f"RMSE: {np.sqrt(mean_squared_error(y_true, y_pred)):.4f}\n")

evaluar_regresion("Ridge Regression", y_test_reg, y_pred_ridge)
evaluar_regresion("Random Forest Regressor", y_test_reg, y_pred_rfr)

# 4. Gráfico y Feature Importance del mejor modelo (Random Forest)
print("Generando Gráfico de Dispersión y extrayendo Importancia de Variables...")

plt.figure(figsize=(8, 6))
plt.scatter(y_test_reg, y_pred_rfr, alpha=0.6, color='#4C72B0', edgecolor='k')
plt.plot([y_test_reg.min(), y_test_reg.max()], [y_test_reg.min(), y_test_reg.max()], 'r--', lw=2)
plt.title("Predicción vs Realidad: Puntaje de Satisfacción (Random Forest)", fontweight='bold')
plt.xlabel("Puntaje Real")
plt.ylabel("Puntaje Predicho")
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()

# Importancia de Variables
importancias = best_rfr.feature_importances_
top5_idx = np.argsort(importancias)[-5:][::-1]
nombres_features = X_train.columns

print("--- TOP 5 VARIABLES MÁS IMPORTANTES (Satisfacción) ---")
for i in top5_idx:
    print(f"{nombres_features[i]}: {importancias[i]:.4f}")
```

## Parte 4: Procesamiento con Apache Spark MLlib

```python
print("======================================================")
print(" PARTE 4: PROCESAMIENTO CON APACHE SPARK MLLIB")
print("======================================================\n")

from pyspark.sql import SparkSession # type: ignore
from pyspark.ml import Pipeline # type: ignore
from pyspark.ml.feature import StringIndexer, VectorAssembler # type: ignore
from pyspark.ml.classification import RandomForestClassifier # type: ignore
from pyspark.ml.evaluation import BinaryClassificationEvaluator, MulticlassClassificationEvaluator # type: ignore
import scipy.stats as stats
import pandas as pd
import numpy as np

# 1. Inicializar Spark
spark = SparkSession.builder.appName("EduTechAnalytics").getOrCreate()

# 2. Cargar datos en Spark
df_spark = spark.read.csv('02. Material de apoyo - Estudiantes_edutech.csv', header=True, inferSchema=True)

# Crear la columna label binaria (se renombra curso_completado)
df_spark = df_spark.withColumnRenamed("curso_completado", "label")

# 3. Preparación: StringIndexer y VectorAssembler
columnas_categoricas = ['genero', 'pais', 'dispositivo_principal', 'horario_estudio', 'nivel_educacion', 'experiencia_previa', 'plataforma_origen']
columnas_numericas = ['edad', 'horas_estudio_semanal', 'sesiones_totales', 'tareas_completadas', 'tareas_totales', 'participacion_foros', 'promedio_evaluaciones', 'monto_inversion']

# Crear los indexadores dinámicamente
indexadores = [StringIndexer(inputCol=c, outputCol=c+"_index", handleInvalid="keep") for c in columnas_categoricas]

# Ensamblar todas las variables (índices categóricos + numéricas)
columnas_ensamblador = [c+"_index" for c in columnas_categoricas] + columnas_numericas
ensamblador = VectorAssembler(inputCols=columnas_ensamblador, outputCol="features")

# 4. Modelo y Pipeline
rf_spark = RandomForestClassifier(featuresCol="features", labelCol="label", seed=42)
pipeline = Pipeline(stages=indexadores + [ensamblador, rf_spark])

# 5. División (70/30) y Entrenamiento
train_spark, test_spark = df_spark.randomSplit([0.7, 0.3], seed=42)
modelo_spark = pipeline.fit(train_spark)

# 6. Predicciones y Evaluación
predicciones_spark = modelo_spark.transform(test_spark)

eval_roc = BinaryClassificationEvaluator(labelCol="label", rawPredictionCol="rawPrediction", metricName="areaUnderROC")
eval_multi = MulticlassClassificationEvaluator(labelCol="label", predictionCol="prediction")

roc_auc_spark = eval_roc.evaluate(predicciones_spark)
acc_spark = eval_multi.evaluate(predicciones_spark, {eval_multi.metricName: "accuracy"})
f1_spark = eval_multi.evaluate(predicciones_spark, {eval_multi.metricName: "f1"})

print(f"Resultados Spark - ROC-AUC: {roc_auc_spark:.4f}")
print(f"Resultados Spark - Accuracy: {acc_spark:.4f}")
print(f"Resultados Spark - F1-Score: {f1_spark:.4f}\n")

print("Muestra de Predicciones en Spark:")
predicciones_spark.select("label", "prediction", "probability").show(5, truncate=False)
```

## Parte 5: Inferencia Estadística

```python
print("======================================================")
print(" PARTE 5: INFERENCIA ESTADÍSTICA")
print("======================================================\n")

# Para la inferencia, usar el DataFrame original de Pandas
satisfaccion = df['puntaje_satisfaccion']

# 5.1 Intervalo de Confianza al 95% para la media
media_sat = np.mean(satisfaccion)
error_estandar = stats.sem(satisfaccion)
intervalo_confianza = stats.t.interval(0.95, len(satisfaccion)-1, loc=media_sat, scale=error_estandar)

print(f"1. Intervalo de Confianza (95%) para el Puntaje de Satisfacción:")
print(f"   La media poblacional se encuentra entre {intervalo_confianza[0]:.2f} y {intervalo_confianza[1]:.2f} (Media muestral: {media_sat:.2f})\n")

# 5.2 Prueba de Hipótesis (T-test)
# H0: No hay diferencia en el puntaje promedio de satisfacción (Media Completados = Media No Completados)
# H1: Sí hay diferencia significativa (Media Completados != Media No Completados)
sat_completados = df[df['curso_completado'] == 1]['puntaje_satisfaccion']
sat_no_completados = df[df['curso_completado'] == 0]['puntaje_satisfaccion']

estadistico_t, valor_p = stats.ttest_ind(sat_completados, sat_no_completados, equal_var=False)

print(f"2. Prueba de Hipótesis (Completados vs No Completados):")
print(f"   Estadístico t: {estadistico_t:.4f}")
print(f"   Valor p: {valor_p:.4e}")

alfa = 0.05
if valor_p < alfa:
    print(f"   Conclusión: Rechazamos H0 (p < 0.05). Existe una diferencia estadísticamente significativa en la satisfacción entre quienes completan el curso y quienes no.")
else:
    print(f"   Conclusión: No rechazamos H0 (p >= 0.05). No hay evidencia suficiente para afirmar que la satisfacción es diferente.")
```
